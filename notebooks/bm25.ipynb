# BM25

BM25 is a algorithm that ranks the query terms the "match" the words in the corpus. This is one of the most used retrieval algorithms and one of the most simple. BM25 Retrieval is sparse, meaning, it is a word-based search. BM25, which is sparse retrieval, ensures exact matching between query and corpus.


## Load in documents and create corpus

If using ```from rag.documents import load_documents``` make sure that the page content is present for the corpus when using BM25. BM25 does not have to be chunked.

````python
from rag.documents import load_documents

FILE_PATH = "/home/nick/github-projects/Sec-Rag/data/google_10K.pdf"

docs = load_documents(FILE_PATH)

corpus = [doc.page_content for doc in docs]
print(f"Length of corpus: {len(corpus)}")
````

## Tokenize corpus and initialize retriever.

Tokenize the corpus so the tokens can be retrieved from BM25 retriever.

````python
from rag.bm25 import BM25Retriever, tokenize

print(tokenize(corpus[0])[:20])

retriever = BM25Retriever()
retriever.add_documents(docs)
retriever
````

## Create Query and then tokenized the query.

The query has to be tokenized. This makes it so the tokens match what is in BM25 Retriever.

````python
query = "how much did Google spend on research and development in 2025"

print(tokenize(query))

results = retriever.retrieve(query, top_k=5)

for doc in results:
    print(f"\n[page {doc.metadata['page']}]\n{doc.page_content[:200]}")
````
